```python
def deutsch_jozsa_circuit(oracle):
    qc = QuantumCircuit(2, 1)
    
    # Prepare the initial state
    qc.h([0,1])  # 모든 큐비트를 중첩 상태로
    qc.x(1)      # 보조 큐비트를 |1>으로
    qc.h(1)      # Hadamard 적용
    
    # 오라클 적용
    qc.append(oracle, [0,1])
    
    # 역방향 Hadamard
    qc.h(0)
    
    # 측정
    qc.measure(0, 0)
    return qc
```
